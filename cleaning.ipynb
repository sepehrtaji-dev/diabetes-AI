```python
import pandas as pd
df = pd.read_csv("diabetes_prediction_dataset.csv")
df.head()
```

```python
df.drop_duplicates(inplace=True)
df.dropna()
```

```python
df['gender'] = df['gender'].astype('float32')
df['smoking_history'] = df['smoking_history'].astype('category')
df['hypertension'] = df['hypertension'].astype('int8')
df['heart_disease'] = df['heart_disease'].astype('int8')
df['diabetes'] = df['diabetes'].astype('int8')
df['age'] = df['age'].astype('float32')
df['bmi'] = df['bmi'].astype('float32')
df['HbA1c_level'] = df['HbA1c_level'].astype('float32')
df['blood_glucose_level'] = df['blood_glucose_level'].astype('float32')
```

```python
def count_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return ((data[column] < lower) | (data[column] > upper)).sum()
```

```python
for col in ['age', 'bmi', 'HbA1c_level', 'blood_glucose_level']:
    print(f'{col}: {count_outliers_iqr(df, col)} outliers')

df.to_csv('diabetes_prediction_dataset_cleaned.csv', index=False)

print('Cleaning complete. Cleaned file saved.')
```

```python
gender_mapping = {'Female': 0, 'Male': 1, 'Other': 2}
df['gender'] = df['gender'].map(gender_mapping)

smoking_mapping = {
    'never': 0,
    'current': 1,
    'former': 2,
    'ever': 3,
    'not current': 4
}
df['smoking_history'] = df['smoking_history'].map(smoking_mapping).fillna(-1).astype('int8')
```

```python
df.to_csv('diabetes_prediction_dataset_cleaned.csv', index=False)
```
